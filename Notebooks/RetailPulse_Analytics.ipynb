```python
reference_date = pd.Timestamp("2025-12-31")

customer_analysis["Recency"] = (
    reference_date - customer_analysis["Last_Purchase"]
).dt.days

display(
    customer_analysis[
        ["Customer_ID", "Recency", "Total_Orders", "Total_Spending"]
    ].head(10)
)
```

```python
customer_analysis["R_Score"] = pd.qcut(
    customer_analysis["Recency"],
    q=5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

customer_analysis["F_Score"] = pd.qcut(
    customer_analysis["Total_Orders"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

customer_analysis["M_Score"] = pd.qcut(
    customer_analysis["Total_Spending"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

display(customer_analysis[
    ["Customer_ID", "Recency", "Total_Orders",
     "Total_Spending", "R_Score", "F_Score", "M_Score"]
].head(10))
```

```python
def segment_customer(row):

    r = row["R_Score"]
    f = row["F_Score"]
    m = row["M_Score"]

    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"

    elif r >= 3 and f >= 4:
        return "Loyal Customers"

    elif r >= 4 and f <= 2:
        return "New / Recent Customers"

    elif r <= 2 and f >= 3:
        return "At Risk"

    else:
        return "Hibernating"


customer_analysis["Customer_Segment"] = (
    customer_analysis.apply(segment_customer, axis=1)
)

display(customer_analysis.head(15))
```

```python
segment_summary = customer_analysis.groupby(
    "Customer_Segment"
).agg(
    Total_Customers=("Customer_ID", "count"),
    Average_Orders=("Total_Orders", "mean"),
    Average_Spending=("Total_Spending", "mean"),
    Total_Revenue=("Total_Spending", "sum")
).reset_index()

display(segment_summary.round(2))
```

```python
# Independent variables (Features)
X = churn[
    [
        "Total_Orders",
        "Total_Quantity",
        "Total_Revenue",
        "Average_Order_Value",
        "Customer_Tenure_Days"
    ]
]

# Dependent variable (Target)
y = churn["Churn_Status"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)

display(X.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())
```

```python
inventory["Reorder_Status"] = inventory.apply(
    lambda row: "Reorder Required"
    if row["Stock_Level"] <= row["Reorder_Level"]
    else "Stock Sufficient",
    axis=1
)

print(inventory["Reorder_Status"].value_counts())
```

```python
import matplotlib.pyplot as plt

risk_counts = inventory["Reorder_Status"].value_counts()

risk_counts.plot(kind="bar", figsize=(8, 5))

plt.title("Inventory Reorder Status")
plt.xlabel("Reorder Status")
plt.ylabel("Number of Products")
plt.xticks(rotation=0)
plt.show()
```

```python
# Calculate target stock
inventory["Target_Stock"] = (
    inventory["Reorder_Level"] + inventory["Safety_Stock"]
)

# Calculate reorder quantity
inventory["Reorder_Quantity"] = (
    inventory["Target_Stock"] - inventory["Stock_Level"]
).clip(lower=0)

# Display products that need reorder
reorder_products = inventory[
    inventory["Reorder_Status"] == "Reorder Required"
].copy()

display(
    reorder_products[
        [
            "Product_ID",
            "Category",
            "Stock_Level",
            "Reorder_Level",
            "Safety_Stock",
            "Target_Stock",
            "Reorder_Quantity",
            "Lead_Time_Days"
        ]
    ].head(10)
)
```

```python
# Calculate estimated purchasing cost for each product
reorder_products["Estimated_Reorder_Cost"] = (
    reorder_products["Reorder_Quantity"] *
    reorder_products["Unit_Cost"]
)

# Total reorder units
total_reorder_units = reorder_products["Reorder_Quantity"].sum()

# Total purchasing cost
total_reorder_cost = reorder_products["Estimated_Reorder_Cost"].sum()

print("Products Requiring Reorder:", len(reorder_products))

print("Total Reorder Units:", total_reorder_units)

print("Estimated Purchasing Cost:",
      round(total_reorder_cost, 2))
```

```python
category_reorder = reorder_products.groupby("Category").agg(
    Products_To_Reorder=("Product_ID", "count"),
    Total_Units=("Reorder_Quantity", "sum"),
    Total_Cost=("Estimated_Reorder_Cost", "sum")
).reset_index()

category_reorder = category_reorder.sort_values(
    by="Total_Cost",
    ascending=False
)

display(category_reorder)
```

```python
demand = files["daily_product_demand"]

print("Shape:", demand.shape)

print("\nColumns:")
print(demand.columns.tolist())

print("\nData Types:")
print(demand.dtypes)

print("\nFirst 5 rows:")
display(demand.head())
```

```python
print("Start Date:", demand["Order_Date"].min())
print("End Date:", demand["Order_Date"].max())

print("\nUnique Products:",
      demand["Product_ID"].nunique())
```

```python
# Total observation days
total_days = (
    demand["Order_Date"].max() -
    demand["Order_Date"].min()
).days + 1

print("Total Observation Days:", total_days)

# Product-wise total demand
product_demand = demand.groupby("Product_ID").agg(
    Total_Units_Sold=("Units_Sold", "sum"),
    Total_Orders=("Orders", "sum"),
    Total_Revenue=("Daily_Revenue", "sum")
).reset_index()

# Average daily demand
product_demand["Average_Daily_Demand"] = (
    product_demand["Total_Units_Sold"] / total_days
)

display(product_demand.head())
```

```python
inventory_analysis = inventory.merge(
    product_demand,
    on="Product_ID",
    how="left"
)

print("Merged Shape:", inventory_analysis.shape)

display(inventory_analysis.head())
```

```python
inventory_analysis["Stock_Cover_Days"] = (
    inventory_analysis["Stock_Level"] /
    inventory_analysis["Average_Daily_Demand"]
)

inventory_analysis["Lead_Time_Demand"] = (
    inventory_analysis["Average_Daily_Demand"] *
    inventory_analysis["Lead_Time_Days"]
)

# Dynamic Reorder Point
inventory_analysis["Dynamic_Reorder_Point"] = (
    inventory_analysis["Lead_Time_Demand"] +
    inventory_analysis["Safety_Stock"]
)

display(
    inventory_analysis[
        [
            "Product_ID",
            "Stock_Level",
            "Average_Daily_Demand",
            "Stock_Cover_Days",
            "Lead_Time_Days",
            "Lead_Time_Demand",
            "Safety_Stock",
            "Dynamic_Reorder_Point"
        ]
    ].head(10).round(2)
)
```

```python
# Dynamic inventory classification

inventory_analysis["Dynamic_Reorder_Status"] = inventory_analysis.apply(
    lambda row: "Reorder Required"
    if row["Stock_Level"] <= row["Dynamic_Reorder_Point"]
    else "Stock Sufficient",
    axis=1
)

print(
    inventory_analysis["Dynamic_Reorder_Status"].value_counts()
)
```

```python
# Compare both reorder approaches

comparison = pd.crosstab(
    inventory_analysis["Reorder_Status"],
    inventory_analysis["Dynamic_Reorder_Status"]
)

display(comparison)
```

```python
# Calculate stock buffer after accounting for lead time demand

inventory_analysis["Stock_Buffer"] = (
    inventory_analysis["Stock_Level"] -
    inventory_analysis["Lead_Time_Demand"]
)

# Identify products with insufficient stock during lead time

urgent_products = inventory_analysis[
    inventory_analysis["Stock_Buffer"] < 0
].copy()

urgent_products = urgent_products.sort_values(
    by="Stock_Buffer"
)

display(
    urgent_products[
        [
            "Product_ID",
            "Category",
            "Stock_Level",
            "Average_Daily_Demand",
            "Lead_Time_Days",
            "Lead_Time_Demand",
            "Stock_Buffer",
            "Dynamic_Reorder_Point"
        ]
    ]
)
```

```python
# Dynamic reorder quantity
inventory_analysis["Dynamic_Reorder_Quantity"] = (
    inventory_analysis["Dynamic_Reorder_Point"]
    - inventory_analysis["Stock_Level"]
).clip(lower=0)

# Estimated reorder cost
inventory_analysis["Dynamic_Reorder_Cost"] = (
    inventory_analysis["Dynamic_Reorder_Quantity"]
    * inventory_analysis["Unit_Cost"]
)

# Filter products requiring reorder
dynamic_reorder_products = inventory_analysis[
    inventory_analysis["Dynamic_Reorder_Status"] == "Reorder Required"
].copy()

# Sort by reorder quantity
dynamic_reorder_products = dynamic_reorder_products.sort_values(
    by="Dynamic_Reorder_Quantity",
    ascending=False
)

display(dynamic_reorder_products[
    [
        "Product_ID",
        "Category",
        "Stock_Level",
        "Dynamic_Reorder_Point",
        "Dynamic_Reorder_Quantity",
        "Unit_Cost",
        "Dynamic_Reorder_Cost"
    ]
])
```

```python
total_dynamic_reorder_units = (
    dynamic_reorder_products["Dynamic_Reorder_Quantity"].sum()
)

total_dynamic_reorder_cost = (
    dynamic_reorder_products["Dynamic_Reorder_Cost"].sum()
)

print("Products to Reorder:",
      len(dynamic_reorder_products))

print("Total Reorder Quantity:",
      round(total_dynamic_reorder_units, 2))

print("Total Dynamic Purchasing Cost: ₹",
      round(total_dynamic_reorder_cost, 2))
```

```python
category_reorder_summary = (
    dynamic_reorder_products
    .groupby("Category")
    .agg(
        Products_To_Reorder=("Product_ID", "count"),
        Total_Reorder_Quantity=(
            "Dynamic_Reorder_Quantity", "sum"
        ),
        Total_Purchasing_Cost=(
            "Dynamic_Reorder_Cost", "sum"
        )
    )
    .reset_index()
    .sort_values(
        by="Total_Purchasing_Cost",
        ascending=False
    )
)

display(category_reorder_summary)
```

```python
import numpy as np

def croston_sba(data, alpha=0.1, forecast_periods=30):

    y = data["y"].values

    # Find days with positive demand
    non_zero_indices = np.where(y > 0)[0]

    if len(non_zero_indices) == 0:
        return np.zeros(forecast_periods)

    # Initial demand size and interval
    first_index = non_zero_indices[0]

    demand_size = y[first_index]
    demand_interval = first_index + 1
    last_demand_index = first_index

    # Update estimates whenever demand occurs
    for index in non_zero_indices[1:]:

        interval = index - last_demand_index

        demand_size = (
            alpha * y[index] +
            (1 - alpha) * demand_size
        )

        demand_interval = (
            alpha * interval +
            (1 - alpha) * demand_interval
        )

        last_demand_index = index

    # SBA adjustment
    forecast_value = (
        (1 - alpha / 2) *
        demand_size / demand_interval
    )

    return np.full(forecast_periods, forecast_value)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

def croston_sba_forecast(y, alpha=0.1, periods=30):

    y = np.asarray(y)
    non_zero = np.where(y > 0)[0]

    if len(non_zero) == 0:
        return np.zeros(periods)

    first = non_zero[0]
    demand_size = y[first]
    interval = first + 1
    last = first

    for idx in non_zero[1:]:
        gap = idx - last

        demand_size = (
            alpha * y[idx] +
            (1 - alpha) * demand_size
        )

        interval = (
            alpha * gap +
            (1 - alpha) * interval
        )

        last = idx

    forecast = (1 - alpha / 2) * demand_size / interval

    return np.repeat(forecast, periods)


validation_results = []

for product_id, data in all_product_forecasts.items():

    y = data["y"].values

    for fold in range(3):

        test_end = len(y) - (2 - fold) * 30
        test_start = test_end - 30

        train = y[:test_start]
        actual = y[test_start:test_end]

        # Moving Average
        moving_average = np.repeat(
            np.mean(train[-28:]), 30
        )

        # Croston-SBA
        croston = croston_sba_forecast(
            train, periods=30
        )

        model_predictions = {
            "Moving Average": moving_average,
            "Croston-SBA": croston
        }

        # Prophet
        train_df = data.iloc[:test_start].copy()

        model = Prophet(
            yearly_seasonality=False,
            weekly_seasonality=True,
            daily_seasonality=False
        )

        model.fit(train_df)

        future = model.make_future_dataframe(
            periods=30, freq="D"
        )

        prophet_forecast = model.predict(future)

        model_predictions["Prophet"] = (
            prophet_forecast.tail(30)["yhat"]
            .clip(lower=0).values
        )

        for model_name, prediction in model_predictions.items():

            mae = mean_absolute_error(actual, prediction)

            rmse = np.sqrt(
                mean_squared_error(actual, prediction)
            )

            wape = (
                np.abs(actual - prediction).sum()
                / actual.sum() * 100
                if actual.sum() > 0 else np.nan
            )

            validation_results.append({
                "Product_ID": product_id,
                "Fold": fold + 1,
                "Model": model_name,
                "MAE": mae,
                "RMSE": rmse,
                "WAPE": wape
            })

validation_df = pd.DataFrame(validation_results)

print("Total model evaluations:", len(validation_df))
display(validation_df.head(10).round(3))
```

```python
from prophet import Prophet
import numpy as np
import pandas as pd

final_forecasts = []

for _, row in best_model_per_product.iterrows():

    product_id = row["Product_ID"]
    model_name = row["Model"]

    data = all_product_forecasts[product_id]

    # Selected model ke according forecast
    if model_name == "Croston-SBA":

        predictions = croston_sba_forecast(
            data["y"].values,
            periods=30
        )

    elif model_name == "Moving Average":

        average_demand = data["y"].tail(28).mean()
        predictions = np.repeat(average_demand, 30)

    elif model_name == "Prophet":

        model = Prophet(
            yearly_seasonality=False,
            weekly_seasonality=True,
            daily_seasonality=False
        )

        model.fit(data)

        future = model.make_future_dataframe(
            periods=30,
            freq="D"
        )

        forecast = model.predict(future)

        predictions = (
            forecast.tail(30)["yhat"]
            .clip(lower=0)
            .values
        )

    forecast_dates = pd.date_range(
        start=data["ds"].max() + pd.Timedelta(days=1),
        periods=30,
        freq="D"
    )

    product_forecast = pd.DataFrame({
        "Product_ID": product_id,
        "Selected_Model": model_name,
        "Forecast_Date": forecast_dates,
        "Predicted_Demand": predictions
    })

    final_forecasts.append(product_forecast)

final_forecast_all = pd.concat(
    final_forecasts,
    ignore_index=True
)

print("Total forecast rows:", len(final_forecast_all))
print("Products forecasted:", final_forecast_all["Product_ID"].nunique())

display(final_forecast_all.head(10))
```

```python
forecast_summary_all = final_forecast_all.groupby(
    ["Product_ID", "Selected_Model"]
).agg(
    Total_Forecast_Demand=("Predicted_Demand", "sum"),
    Average_Daily_Forecast=("Predicted_Demand", "mean"),
    Maximum_Daily_Forecast=("Predicted_Demand", "max")
).reset_index()

forecast_summary_all["Recommended_30_Day_Units"] = (
    np.ceil(forecast_summary_all["Total_Forecast_Demand"])
).astype(int)

print("Products:", len(forecast_summary_all))

display(
    forecast_summary_all.sort_values(
        "Total_Forecast_Demand",
        ascending=False
    ).head(10).round(2)
)
```

```python
# Add lead time and inventory information to forecast
forecast_with_inventory = final_forecast_all.merge(
    inventory_analysis[
        [
            "Product_ID",
            "Stock_Level",
            "Safety_Stock",
            "Lead_Time_Days",
            "Unit_Cost"
        ]
    ],
    on="Product_ID",
    how="left"
)

# Arrange forecast dates product-wise
forecast_with_inventory = forecast_with_inventory.sort_values(
    ["Product_ID", "Forecast_Date"]
)

# Identify each product's forecast day
forecast_with_inventory["Forecast_Day"] = (
    forecast_with_inventory.groupby("Product_ID").cumcount() + 1
)

# Keep demand only during supplier lead time
lead_time_demand = forecast_with_inventory[
    forecast_with_inventory["Forecast_Day"]
    <= forecast_with_inventory["Lead_Time_Days"]
]

# Calculate predicted demand during lead time
lead_time_summary = lead_time_demand.groupby(
    "Product_ID"
).agg(
    Forecast_Lead_Time_Demand=(
        "Predicted_Demand", "sum"
    )
).reset_index()

# Merge lead-time demand with product-level forecast
final_inventory_plan = forecast_summary_all.merge(
    inventory_analysis[
        [
            "Product_ID",
            "Stock_Level",
            "Safety_Stock",
            "Lead_Time_Days",
            "Unit_Cost"
        ]
    ],
    on="Product_ID",
    how="left"
).merge(
    lead_time_summary,
    on="Product_ID",
    how="left"
)

# Reorder point = lead-time demand + safety stock
final_inventory_plan["Reorder_Point"] = np.ceil(
    final_inventory_plan["Forecast_Lead_Time_Demand"]
    + final_inventory_plan["Safety_Stock"]
).astype(int)

# Calculate recommended purchase quantity
final_inventory_plan["Recommended_Reorder_Quantity"] = np.ceil(
    (
        final_inventory_plan["Reorder_Point"]
        - final_inventory_plan["Stock_Level"]
    ).clip(lower=0)
).astype(int)

# Estimated purchasing cost
final_inventory_plan["Estimated_Reorder_Cost"] = (
    final_inventory_plan["Recommended_Reorder_Quantity"]
    * final_inventory_plan["Unit_Cost"]
)

final_inventory_plan["Inventory_Status"] = np.where(
    final_inventory_plan["Recommended_Reorder_Quantity"] > 0,
    "Reorder Required",
    "Stock Sufficient"
)

print("Inventory plans:", len(final_inventory_plan))

display(
    final_inventory_plan[
        [
            "Product_ID",
            "Selected_Model",
            "Total_Forecast_Demand",
            "Stock_Level",
            "Forecast_Lead_Time_Demand",
            "Safety_Stock",
            "Reorder_Point",
            "Recommended_Reorder_Quantity",
            "Estimated_Reorder_Cost",
            "Inventory_Status"
        ]
    ].head(15).round(2)
)
```
